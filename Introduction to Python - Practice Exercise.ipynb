```python
#Task 3: Basic Operations
x = 17
y = 8
# Arithmetic
print("Addition:", x + y)
print("Subtraction:", x - y)
print("Multiplication:", x * y)
print("Division:", x / y)
print("Modulus:", x % y)
print("Exponent:", x ** y)

# Comparison
print("x > y:", x > y)
print("x < y:", x < y)
print("x == y:", x == y)
print("x != y:", x != y)
print("x >= y:", x >= y)
print("x <= y:", x <= y)

# Logical
print("x > 10 and y < 5:", x > 10 and y < 5)
print("x < 10 or y > 5:", x < 10 or y > 10)
print("not(x == y):", not(x == y))
```
